```python
import pandas as pd

def load_and_preprocess_data(file_path):
    # Load dataset
    df = pd.read_csv(file_path)
    
    # Fill missing values without using inplace=True
    df['Cuisines'] = df['Cuisines'].fillna('Unknown')
    
    # Convert data types if necessary (e.g., ensuring numeric columns)
    df['Average Cost for two'] = pd.to_numeric(df['Average Cost for two'], errors='coerce')
    df['Votes'] = pd.to_numeric(df['Votes'], errors='coerce')
    
    # Handle missing data (drop rows where important columns are missing)
    df.dropna(subset=['Restaurant ID'], inplace=True)
    
    return df

# Path to your new cleaned data file
file_path = r'C:\Users\DELL\Documents\zomato dataset\zomato_cleaned_data.csv'
df = load_and_preprocess_data(file_path)
```

```python
# Summary statistics of numeric columns
print(df.describe())
```

```python
# Select only numeric columns
numeric_columns = df.select_dtypes(include=['int64', 'float64']).columns

# Create a correlation matrix using only numeric columns
correlation_matrix = df[numeric_columns].corr()

# Plot the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Descriptive statistics
print(df.describe())

# Box plot for numeric columns to detect outliers
plt.figure(figsize=(10, 6))
sns.boxplot(data=df[numeric_columns])
plt.title('Box plot of Numeric Features')
plt.show()
```

```python
# Distribution of 'Aggregate rating'
plt.figure(figsize=(10, 6))
sns.histplot(df['Aggregate rating'], kde=True)
plt.title('Distribution of Aggregate Ratings')
plt.show()

# Scatter plot of 'Votes' vs 'Aggregate rating'
plt.figure(figsize=(10, 6))
sns.scatterplot(x=df['Votes'], y=df['Aggregate rating'])
plt.title('Votes vs Aggregate Rating')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Create some sample data
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Set up the plot
fig, ax = plt.subplots()
line, = ax.plot(x, y)

# Function to update the plot at each frame
def update(frame):
    line.set_ydata(np.sin(x + frame / 10))  # Update the y-data for the line plot
    return line,

# Create the animation
ani = FuncAnimation(fig, update, frames=100, interval=50, blit=True)

# Display the animation
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation

# Create a bar chart with sample data
categories = ['Category A', 'Category B', 'Category C', 'Category D']
values = np.random.randint(1, 10, size=4)

fig, ax = plt.subplots()
bars = ax.bar(categories, values)

# Update function for the bar chart
def update(frame):
    new_values = np.random.randint(1, 10, size=4)  # Random data change
    for bar, new_height in zip(bars, new_values):
        bar.set_height(new_height)
    return bars

# Animate the bar chart
ani = FuncAnimation(fig, update, frames=50, interval=500, blit=False)

# Show the plot
plt.show()
```
